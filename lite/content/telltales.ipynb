# Tell-tales

The tell-tales of anywidget-instruments-automotive, in the browser. Each takes its colour from its function — red for a danger, amber for a warning, green for a function on, blue for the main beam — and shows its name as text.

> **Not vehicle instruments.** See the [safety notice](https://anywidgetinstruments.github.io/anywidget-instruments-automotive/safety/).

```python
# JupyterLite: install the packages (bundled with this site) in the browser kernel
%pip install -q anywidget-instruments-automotive
```

```python
import anywidget_instruments_automotive as aa

cluster = aa.TellTaleCluster(
    [
        ("turn_left", "off"),
        ("low_beam", "on"),
        ("engine", "off"),
        ("oil_pressure", "off"),
        ("high_beam", "off"),
    ],
    label="Tell-tales",
)
cluster
```

Light the engine tell-tale and blink the direction indicator: the lit ones move to the front, red first, then amber, then green and blue.

```python
cluster.set_telltale("engine", "on")
cluster.set_telltale("turn_left", "blinking")
```

A tell-tale on its own. With no state it says so, rather than looking unlit.

```python
oil = aa.TellTale("oil_pressure")
oil
```

```python
oil.state = "on"
```
